```python
import os
import six
import SimpleITK as sitk

from pathlib import Path
from radiomics import featureextractor
```

```python
extractor = featureextractor.RadiomicsFeatureExtractor()

print('Extraction parameters:\n\t', extractor.settings)
print('Enabled filters:\n\t', extractor.enabledImagetypes)
print('Enabled features:\n\t', extractor.enabledFeatures)
```

```python
CT_DIR  = Path("/content/drive/MyDrive/Colab Notebooks/processed/ct_nifti")
MASK_DIR = Path("/content/drive/MyDrive/Colab Notebooks/processed/mask_nifti")
RADIOMICS_DIR  = Path("/content/drive/MyDrive/Colab Notebooks/radiomics_ct")
os.makedirs(RADIOMICS_DIR, exist_ok=True)
```

```python
def get_patient_id(ct_path: Path) -> str:
    id = ct_path.split('_')[0]
    return id
```

```python
def get_mask_and_id_for_patient() -> dict[str, str]:
    """
    returns the path to the mask for the patient along with the patient ID
    """
    masks = {}
    for mask in os.listdir(MASK_DIR):
        patient_id = get_patient_id(mask)
        if patient_id:
          masks[patient_id] = mask

    return masks
```

```python
masks = get_mask_and_id_for_patient()
for ct_path in os.listdir(CT_DIR):
  patient_id = get_patient_id(ct_path)
  mask_path = masks[patient_id]

  imagePath = os.path.join(CT_DIR, ct_path)
  maskPath = os.path.join(MASK_DIR, mask_path)
  print("imagePath" + imagePath)
  print("maskPath" + maskPath)

  try:
    result = extractor.execute(imagePath, maskPath)

    output_filename = f"{patient_id}_radiomics.txt"
    output_path = os.path.join(RADIOMICS_DIR, output_filename)

    print(f"   > Save results: {output_filename}...")

    with open(output_path, 'w', encoding='utf-8') as f:
      f.write(f"patient_id: {patient_id}\n")

      for key, value in result.items():
           f.write(f"{key}: {value}\n")

    print(f"   > Succes")

  except Exception as e:
    print(f"   --- EROARE la procesarea pacientului {patient_id}: {e}")
```
