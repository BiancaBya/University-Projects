```python
import urllib.request


url = 'https://github.com/lauradiosan/AI-UBB/raw/refs/heads/main/2024-2025/labs/lab02/data/texts.txt'

with urllib.request.urlopen(url) as response:
    text = response.read().decode('utf-8')

print(text)
```

```python
# ex 1 numarul de propozitii din text

import nltk
from nltk.tokenize import sent_tokenize

propozitii = sent_tokenize(text)

print("Numărul de propozitii:", len(propozitii))
```

```python
# ex 2 numarul de cuvinte din text
from nltk.tokenize import word_tokenize

cuvinte = word_tokenize(text)
num_words = len(cuvinte)

print("Numărul de cuvinte:", num_words)
```

```python
# ex 3 numarul de cuvinte diferite din text

import re

words = re.findall(r'\b\w+\b', text.lower())  # Convertim totul la litere mici
unique_words = set(words)
num_unique_words = len(unique_words)

print("Numărul de cuvinte diferite:", num_unique_words)
```

```python
# ex 4 cel mai scurt si cel mai lung cuvant (cuvinte)

cele_mai_lungi_cuv = []
cele_mai_scurte_cuv = []

maxim = 0
minim = 100000000
for cuv in cuvinte:
    if len(cuv) > maxim:
        cele_mai_lungi_cuv = [cuv]
        maxim = len(cuv)
    elif len(cuv) == maxim:
        cele_mai_lungi_cuv.append(cuv)
    if len(cuv) < minim :
        cele_mai_scurte_cuv = [cuv]
        minim = len(cuv)
    elif len(cuv) == minim:
        cele_mai_scurte_cuv.append(cuv)


print(cele_mai_lungi_cuv)
print(cele_mai_scurte_cuv)
```

```python
# ex 5 textul fara diacritice
import unicodedata

def remove_diacritics(text_fct):
    return ''.join(
        c for c in unicodedata.normalize('NFD', text_fct) if unicodedata.category(c) != 'Mn'
    )

text_without_diacritics = remove_diacritics(text)

print(text_without_diacritics)
```

```python
import nltk
from nltk.corpus import wordnet

def find_longest_word(text_fct):
    words_fct = nltk.word_tokenize(text_fct)
    longest_word_fct = max(words_fct, key=len)
    return longest_word_fct

def get_synonyms(word):
    synonyms = set()
    for syn in wordnet.synsets(word):
        for lemma in syn.lemmas():
            synonyms.add(lemma.name()) # lemma aduce cuv la forma lui de baza
    return list(synonyms)

longest_word = find_longest_word(text)

synonyms = get_synonyms(longest_word)

print(f"Cel mai lung cuvânt este: {longest_word}")
print(f"Sinonimele acestuia sunt: {synonyms}")
```

```python
from nltk import word_tokenize, FreqDist
import numpy as np
import matplotlib.pyplot as plt
import urllib.request
# NORMALIZARE

sentences = sent_tokenize(text)

def get_frecquencies(propizitii):
    frecvente = []
    for propozitie in propizitii:
        cuvintee = word_tokenize(propozitie.lower())
        freqw = FreqDist(cuvintee) # FreqDist -> dicționar cu frecvența fiecărui cuvânt.
        frecvente.append(freqw) # "Succes si si inceapa aventura ai AI!" => {'Succes':1, 'si':2, 'ai':3, 'inceapa':1, ''aventura': 1}
    return frecvente


def normalize_word_frequencies(propizitii_normalizat):
    frecvente_normalizate = []

    for propozitie in propizitii_normalizat:
        cuvinte_norm = word_tokenize(propozitie.lower())
        word_freq = FreqDist(cuvinte_norm)  # Numără frecvențele cuvintelor
        total_words = len(cuvinte_norm)

        # frecvență: raportăm fiecare frecvență la numărul total de cuvinte
        normalized_sentence_freq = {word: frecv / total_words for word, frecv in word_freq.items()}

        frecvente_normalizate.append(normalized_sentence_freq)

    return frecvente_normalizate

frequencies = get_frecquencies(sentences)
normalized_frequencies = normalize_word_frequencies(sentences)

for idx, freq in enumerate(normalized_frequencies):
    print(f"Propoziția {idx + 1}: {freq}")


for idx, freq in enumerate(normalized_frequencies):
    fig, axes = plt.subplots(1, 1, figsize=(12, 5))
    axes.set_title(f"Propozitia {idx + 1}")
    axes.plot(freq.keys(), list(freq.values()))

for idx, freq in enumerate(frequencies):
    fig, axes = plt.subplots(1, 1, figsize=(12, 5))
    axes.set_title(f"Propozitia {idx + 1}")
    axes.plot(freq.keys(), list(freq.values()))
```
