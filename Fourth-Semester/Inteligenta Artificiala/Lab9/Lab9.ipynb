```python
import transformers

def generate_poetry(prompt, model_name, max_length=50, temperature=0.7, top_k=50, top_p=0.9):
    model = transformers.GPT2LMHeadModel.from_pretrained(model_name)
    tokenizer = transformers.GPT2Tokenizer.from_pretrained(model_name)


    tokenizer.pad_token = tokenizer.eos_token
    pad_token_id = tokenizer.pad_token_id


    input_ids = tokenizer.encode(prompt, return_tensors="pt")
    attention_mask = (input_ids != pad_token_id).long()


    output = model.generate(  # genereaza textul lipsa
        input_ids=input_ids,
        attention_mask=attention_mask,
        max_length=max_length,
        temperature=temperature,
        top_k=top_k,
        top_p=top_p,
        num_return_sequences=1,
        do_sample=True,
        no_repeat_ngram_size=2,
        pad_token_id=pad_token_id
    )

    generated_text = tokenizer.decode(output[0], skip_special_tokens=True)
    return generated_text

def analyze_parameters(prompt):
    model_names = ["gpt2-medium", "gpt2"]
    temperatures = [0.5, 1.2]
    top_ks = [10, 100]
    top_ps = [0.7, 1.0]
    max_lengths = [30, 80]

    results = []

    for model_name in model_names:
        for temp in temperatures:
            for k in top_ks:
                for p in top_ps:
                    for length in max_lengths:
                        print(f"\nModel: {model_name} | Temperature: {temp} | Top-K: {k} | Top-P: {p} | Length: {length}\n")

                        try:
                            poem = generate_poetry(
                                prompt,
                                model_name=model_name,
                                max_length=length,
                                temperature=temp,
                                top_k=k,
                                top_p=p
                            )
                            print(poem)
                        except Exception as e:
                            print(f"Error generating text: {e}")

                        print("-" * 50)

    return results
prompt = "I am sick and i am tired too, I can admit, I am not fireproof"
analyze_parameters(prompt)
```

```python
prompt="Vreau sa zbor departe, sa nu ma mai intorc"
analyze_parameters(prompt)
```

```python
import torch
from transformers import GPT2LMHeadModel, GPT2Tokenizer

poem = """\
Roșu-i cerul sus,
Alb pe dealuri,
Soarele apune
Peste valuri."""

tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
tokenizer.pad_token = tokenizer.eos_token
model = GPT2LMHeadModel.from_pretrained("gpt2")
model.resize_token_embeddings(len(tokenizer))
model.train()

inputs = tokenizer.encode(poem, return_tensors="pt")
inputs = inputs.to(model.device)
optimizer = torch.optim.Adam(model.parameters(), lr=5e-5)

for epoch in range(3):
    outputs = model(inputs, labels=inputs)
    loss = outputs.loss
    loss.backward()
    optimizer.step()
    optimizer.zero_grad()

model.save_pretrained("gpt2-poetry")
tokenizer.save_pretrained("gpt2-poetry")
```

```python
from transformers import GPT2Tokenizer

def analyze_tokenizer(text, model_name):
    tokenizer = GPT2Tokenizer.from_pretrained(model_name)
    tokens = tokenizer.tokenize(text)
    token_ids = tokenizer.convert_tokens_to_ids(tokens)

    print(f"Textul original: {text}\n")
    print(f"Tokeni generați: {tokens}\n")
    print(f"ID-uri tokeni: {token_ids}\n")


print("English - gpt2:")
analyze_tokenizer("I feel like I'm drowning", "gpt2")

print("\nRomana - gpt2:")
analyze_tokenizer("Stam singur in cavou si era vant", "gpt2")

print("\nEnglish - gpt2-medium:")
analyze_tokenizer("I feel like I'm drowning", "gpt2-medium")

print("\nRomana - gpt2-medium:")
analyze_tokenizer("Stam singur in cavou si era vant", "gpt2-medium")

```

```python
import nltk
from nltk.translate.bleu_score import sentence_bleu

nltk.download('punkt')

def bleu_score(reference, candidate):
    reference_tokens = nltk.word_tokenize(reference.lower())
    candidate_tokens = nltk.word_tokenize(candidate.lower())
    bleu_score = sentence_bleu([reference_tokens], candidate_tokens)
    return bleu_score


text = "There aren't many words that you can say, That could ever get my mind to change"
gpt2 = "There aren't many words that you can say, That could ever get my mind to change"
gpt2_medium = "There aren't many words that you can say, That could ever get my mind to change"

print(f"BLEU Score GPT-2: {bleu_score(text, gpt2)}")
print(f"BLEU Score GPT-2 Medium: {bleu_score(text, gpt2_medium)}")
```
