```python
import os
from PIL import Image

def apply_sepia(input_image: Image.Image) -> Image.Image:

    img = input_image.convert("RGB")
    width, height = img.size
    pixels = img.load()

    for y in range(height):
        for x in range(width):
            r, g, b = pixels[x, y]

            tr = int(0.393 * r + 0.769 * g + 0.189 * b)
            tg = int(0.349 * r + 0.686 * g + 0.168 * b)
            tb = int(0.272 * r + 0.534 * g + 0.131 * b)

            pixels[x, y] = (
                min(255, tr),
                min(255, tg),
                min(255, tb)
            )

    return img

def process_folder(input_dir: str, output_dir: str):

    valid_exts = ('.jpg', '.jpeg', '.png', '.bmp', '.tiff')

    for filename in os.listdir(input_dir):
        if not filename.lower().endswith(valid_exts):
            continue

        input_path = os.path.join(input_dir, filename)
        try:
            with Image.open(input_path) as img:
                sepia_img = apply_sepia(img)
                output_path = os.path.join(output_dir, filename)
                sepia_img.save(output_path)
                print(f"Procesată: {filename}")
        except Exception as e:
            print(f"Nu s-a putut procesa {filename}: {e}")


input_folder = "photos"
output_folder = "photos_sepia"
#process_folder(input_folder, output_folder)



```
